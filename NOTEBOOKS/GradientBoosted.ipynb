Gradient Boosted

```python
import sklearn as skl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn import preprocessing
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold, cross_val_predict
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, make_scorer

df = pd.read_csv('../DATASETS/salario_tratamento.csv', index_col=0)
```

```python
x = df.drop('income', axis=1)
y = df['income']
```

Label Encoding para as colunas categoricas 

```python
categorical_columns = ['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex', 'native.country','age.group']
numeric_columns = ['capital.diff', 'hours.per.week']

label_encoder = LabelEncoder()
for column in categorical_columns:
    x[column] = label_encoder.fit_transform(x[column])
```

Normalização de Dados

```python
scaler = StandardScaler()
x[numeric_columns] = scaler.fit_transform(x[numeric_columns])
```

Criação do Modelo Gradient Boosted Trees

```python
model = GradientBoostingClassifier(n_estimators=100, random_state=2023)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=2023)
```

Treino do Modelo

```python
def calculate_metrics(y_true, y_pred):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, pos_label='>50K')
    recall = recall_score(y_true, y_pred, pos_label='>50K')
    cm = confusion_matrix(y_true, y_pred)
    return accuracy, precision, recall, cm

scorer = make_scorer(accuracy_score)
y_pred_cv = cross_val_predict(model, x, y, cv=skf)
accuracy, precision, recall, cm = calculate_metrics(y, y_pred_cv)
```

Resultados

```python
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("Confusion Matrix:\n", cm)
```
