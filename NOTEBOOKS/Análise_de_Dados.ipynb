```python

import sklearn as skl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn import preprocessing

df = pd.read_csv('../DATASETS/salario.csv')
```

Histograma para as Idades

```python
plt.hist(df['age'], bins=10, color='blue', edgecolor='black')
plt.xlabel('Idade')
plt.ylabel('Indivíduos')
plt.title('Histograma das Idades')
```

Gráfico de Barras para Workclass


```python
workclass_count=df['workclass'].value_counts()

workclass_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Classe de Trabalho')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Workclass')
```

Gráfico de Barras para Educação

```python
education_count=df['education'].value_counts()
education_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Nível de Educação')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Education')
```

Gráfico de Barras para Educação (Númerico)


```python
educationum_count=df['education.num'].value_counts()
educationum_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Nível de Educação (Númerico)')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Education.num')
```

Gráfico de Barras do Estado Matrimonial

```python
marital_count=df['marital.status'].value_counts()
marital_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Estado Matrimonial')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Marital.status')
```

Gráfico de Barras da Ocupação

```python
occupation_count= df['occupation'].value_counts()
occupation_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Ocupação')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Occupation')
```

Gráfico de Barras do Relacionamento

```python
relationship_count=df['relationship'].value_counts()
relationship_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Relacionamento')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Relationship')
```

Gráfico de barras para a Raça

```python
race_count=df['race'].value_counts()
race_count.plot(kind='bar', color='blue', edgecolor='black')
plt.xlabel('Raça')
plt.ylabel('Nº Indivíduos')
plt.title('Gráfico de Barras da Raça')
```

Pie Chart para o Sexo

```python
sex_count=df['sex'].value_counts()
plt.pie(sex_count, labels=sex_count.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral'])
plt.title('Pie chart do Sexo')
```

Histograma para Hours-per-Week

```python
plt.hist(df['hours.per.week'], bins=10, color='blue', edgecolor='black')
plt.xlabel('Horas')
plt.ylabel('Indivíduos')
plt.title('Histograma das Horas por Semana')
```

Pie Chart para o Income

```python
income_count = df['income'].value_counts()
plt.pie(income_count, labels=income_count.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral'])
plt.title('Distribuição de Income')
```
