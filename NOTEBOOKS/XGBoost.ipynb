```python
import sklearn as skl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn import preprocessing
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold, cross_val_predict
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, make_scorer
import xgboost as xgb

df = pd.read_csv('../DATASETS/salario_tratamento.csv', index_col=0)
```

```python
x = df.drop('income', axis=1)
y = df['income']
```

Label Encodig

```python
categorical_columns = ['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex', 'native.country','age.group']
numeric_columns = ['capital.diff', 'hours.per.week']

label_encoder = LabelEncoder()
for column in categorical_columns:
    x[column] = label_encoder.fit_transform(x[column])
```

Income necessita de ser Encoded também

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

Normalização dos dados

```python
scaler = StandardScaler()
x[numeric_columns] = scaler.fit_transform(x[numeric_columns])
```

Treino e Criação do modelo XGBoost

```python
model = xgb.XGBClassifier(objective='binary:logistic', random_state=42)
skf= StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

Treino do Modelo

```python
def calculate_metrics(y_true, y_pred):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred)
    return accuracy, precision, recall, cm

scorer = make_scorer(accuracy_score)
y_pred_cv = cross_val_predict(model, x, y_encoded, cv=skf)
accuracy, precision, recall, cm = calculate_metrics(y_encoded, y_pred_cv)
```

Resultados

```python
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("Confusion Matrix:\n", cm)
```
