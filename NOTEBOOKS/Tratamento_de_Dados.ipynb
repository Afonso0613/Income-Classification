## Tratamento de Dados

```python

import sklearn as skl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn import preprocessing

df = pd.read_csv('../DATASETS/salario.csv')
```

Trocar todos os dados '?' por missing values

```python
missing_data = df.isnull().sum()
print(missing_data)
```

```python
df.replace('?', np.nan, inplace=True)
```

```python
df['occupation'] = df['occupation'].fillna(df['occupation'].mode()[0])
df['workclass'] = df['workclass'].fillna(df['workclass'].mode()[0])
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])
```

Adicionar nova coluna capital.diff

```python
df['capital.diff']= df['capital.gain']-df['capital.loss']
```

```python
df.drop(['capital.gain', 'capital.loss'], axis=1, inplace=True)
```

Remoção da coluna education.num e fnlwgt

```python
df.drop(['education.num', 'fnlwgt'], axis=1, inplace=True)
```

Binning das idades

```python
bins = [17, 25, 35, 45, 55, 65, 90]
labels = ['17-25', '26-35', '36-45', '46-55', '56-65', '66-90']
df['age.group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```

```python
df.drop(['age'], axis=1, inplace=True)
```
